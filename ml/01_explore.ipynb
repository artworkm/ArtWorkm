```python

import pandas as pd
import numpy as np
import glob
import os

# Locate Parquet files
files = sorted(glob.glob("../data/cicids2017/*.parquet"))

print("Parquet files found:", len(files))

# Check whether files exist
if not files:
    raise FileNotFoundError("No Parquet files found!")

# Read datasets
frames = []

for f in files:
    print("Reading:", os.path.basename(f))

    d = pd.read_parquet(f)

    d.columns = d.columns.str.strip()

    num = d.select_dtypes(include="number").columns
    d[num] = d[num].astype("float32")

    frames.append(d)

# Combine datasets
df = pd.concat(frames, ignore_index=True)

print("\nDataset successfully loaded!")
print("Dataset shape:", df.shape)

# Inspect columns and labels
print("\nDataset columns:")
print(df.columns.tolist())

print("\nFirst five rows:")
display(df.head())
```
